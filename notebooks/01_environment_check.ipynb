# Environment check
Run the code cells in order. This notebook checks the runtime only; it does not load a model or run a safety evaluation.
## Reflection
- What Python version and runtime do you see?
- Is a GPU available? If so, what device and memory does PyTorch report?
- Do the observed resources match the environment you expected?
- What would you want to confirm before beginning a model evaluation?

```python
import sys
print("Python version:", sys.version)
print("Python executable:", sys.executable)
```

```python
try:
    import torch
    print("PyTorch version:", torch.__version__)
    print("CUDA available:", torch.cuda.is_available())
    if torch.cuda.is_available():
        device = torch.cuda.get_device_name(0)
        props = torch.cuda.get_device_properties(0)
        memory_gib = props.total_memory / (1024 ** 3)
        print("GPU:", device)
        print(f"GPU memory: {memory_gib:.2f} GiB")
    else:
        print("No CUDA GPU is available in this runtime.")
except ImportError:
    print("PyTorch is not installed in this runtime.")
```
